Este ejercicio combina las herramientas de las unidades 4a y 4b, reutilizando el corpus UNGDC de la Unidad 3 ([UN General Debate Corpus 1946-2023 — Kaggle](https://www.kaggle.com/datasets/namigabbasov/united-nations-general-debate-corpus-1946-2023)).

**Pasos sugeridos:**

### <img src="https://upload.wikimedia.org/wikipedia/commons/1/1b/R_logo.svg" alt="R" style="height: 1em !important; width: auto !important; max-width: none !important; vertical-align: middle;"/> En R

1. Filtra el corpus UNGDC a dos grupos de tu elección (por ejemplo, un bloque regional específico vs. otro, o una década reciente vs. una antigua).
2. Calcula `textstat_keyness()` entre ambos grupos y expórtalo a CSV con `write_csv()`, tal como se hizo en la Unidad 4a con `tstat_key_movies.csv`.

### <img src="https://upload.wikimedia.org/wikipedia/commons/c/c3/Python-logo-notext.svg" alt="Python" style="height: 1em !important; width: auto !important; max-width: none !important; vertical-align: middle;"/> En Python

3. Carga el CSV con `pandas`, extrae las palabras más asociadas a cada grupo (como en la Unidad 4b), y construye un *prompt* para que un LLM interprete los patrones encontrados.
4. Compara la interpretación del LLM con tu propia lectura de las palabras clave. ¿Coinciden?

**Discusión:**

5. La Unidad 4b mostró que un modelo de 3B parámetros es limitado frente a modelos comerciales grandes. ¿Qué riesgos existen al usar un LLM para interpretar diferencias de vocabulario entre países con sistemas políticos e idiomas de origen distintos? ¿Podría introducir sesgos culturales en su interpretación?
