## Ejercicio 1: Keyness y Dispersión por Género Implícito en Reseñas de Cine <img src="https://upload.wikimedia.org/wikipedia/commons/1/1b/R_logo.svg" alt="R" style="height: 1em !important; width: auto !important; max-width: none !important; vertical-align: middle;"/>

`data_corpus_moviereviews` no tiene una *docvar* de género, así que la crearás tú mismo a partir del texto.

**Pasos sugeridos:**

1. Crea una nueva *docvar* `genre_mention` que clasifique cada reseña según si menciona la palabra **"comedy"** o la palabra **"horror"** en el texto (usa `stringr::str_detect()` o `grepl()` sobre el texto crudo del corpus). Descarta las reseñas que mencionen ambas palabras o ninguna.
2. Tokeniza, construye la DFM, y usa `textstat_keyness()` para comparar el vocabulario entre ambos grupos.
3. Visualiza los resultados con `textplot_keyness()`. ¿Qué palabras resultan más características de cada grupo? ¿Tienen sentido dado el género?
4. Elige una palabra clave (por ejemplo `"funny"` o `"scary"`) y usa `textplot_xray()` para observar su dispersión a lo largo de los documentos de cada grupo.
5. **Discusión**: clasificar documentos según la simple mención de una palabra ("comedy"/"horror") es un método rudimentario. ¿Qué problemas de validez podría tener este enfoque? ¿Cómo lo mejorarías?

> 💡 Tip: reutiliza el código de las secciones "Keyness" y "Dispersión Textual" de esta unidad como plantilla.
