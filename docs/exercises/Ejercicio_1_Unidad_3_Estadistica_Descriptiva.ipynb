## Ejercicio 1: Estadística Descriptiva del UN General Debate Corpus

En este ejercicio aplicarás las herramientas de estadística descriptiva de esta unidad sobre un corpus real y mucho más grande que `data_corpus_inaugural`: el **UN General Debate Corpus (UNGDC)**, más de 10.000 discursos de 202 países ante la Asamblea General de la ONU (1946–2023), disponible en Kaggle bajo licencia **CC0 (dominio público)**:

🔗 [UN General Debate Corpus 1946-2023 — Kaggle](https://www.kaggle.com/datasets/namigabbasov/united-nations-general-debate-corpus-1946-2023)

El dataset contiene, entre otras, las columnas `doc_id` (identificador único), `text` (el discurso), `ccodealp` (código de país de 3 letras), `session` (número de sesión) y `year` (año).

> ⚠️ El dataset completo pesa ~62 MB y contiene miles de documentos. Descárgalo desde Kaggle, colócalo en el mismo directorio que este notebook (ajusta el nombre del archivo en el código según cómo se haya descargado), y trabaja con un **subconjunto manejable** — por ejemplo, un rango de 10 años — para mantener los tiempos de ejecución razonables (ver paso 1).

```r
library(readr)
library(quanteda)

ungdc <- read_csv("un-general-debates.csv")  # ajusta el nombre según cómo se haya descargado el archivo

# Filtra a un subconjunto manejable antes de construir el corpus, p. ej. un rango de 10 años:
# ungdc <- ungdc[ungdc$year >= 2010 & ungdc$year <= 2019, ]

# Crear corpus con docvars equivalentes a data_corpus_inaugural
corp <- corpus(ungdc, text_field = "text",
               docid_field = "doc_id")  # doc_id es el identificador único; "country" no lo es (se repite entre años)
docvars(corp, "Year")    <- ungdc$year
docvars(corp, "Country") <- ungdc$ccodealp  # la columna se llama ccodealp, no "country"
docvars(corp, "Session") <- ungdc$session
```
