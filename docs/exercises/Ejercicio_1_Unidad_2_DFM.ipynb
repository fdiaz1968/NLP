## Ejercicio 1: Selección de Features y su Efecto en la Sparsity

En este ejercicio construirás una DFM sobre un subconjunto distinto de `data_corpus_inaugural`, y explorarás cómo las decisiones de selección de *features* afectan tanto el tamaño del vocabulario como la sparsity de la matriz.

**Pasos sugeridos:**

1. Selecciona los discursos de los **últimos 15 presidentes** del corpus (`tail(data_corpus_inaugural, 15)`), tokeniza (elimina puntuación) y construye la DFM.
2. Calcula la sparsity de la DFM sin ningún filtro.
3. Aplica `dfm_trim()` con al menos dos umbrales distintos de `min_termfreq` (por ejemplo, 3 y 10) y registra, para cada uno: el número de *features* resultante y la sparsity.
4. Usa `topfeatures()` para obtener las 15 palabras más frecuentes según `scheme = "count"` y según `scheme = "docfreq"`. ¿Coinciden las listas? Si no, ¿qué tipo de palabras aparecen en una lista y no en la otra, y por qué?
5. Agrupa la DFM por alguna *docvar* disponible (por ejemplo, `Party`) y compara las palabras más frecuentes entre grupos usando `topfeatures(..., groups = ...)`.
6. **Discusión**: la sección "Una mirada hacia adelante" de esta unidad menciona que los *embeddings* evitan el problema de la sparsity. Basado en lo que observaste al variar `min_termfreq`, ¿en qué situaciones el trade-off entre sparsity y pérdida de información sería más riesgoso al reducir agresivamente el vocabulario?

> 💡 Tip: reutiliza el código de las secciones "Sparsity" y "Selección de Features" como plantilla.
