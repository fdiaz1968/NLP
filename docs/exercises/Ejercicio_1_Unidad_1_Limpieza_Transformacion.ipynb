## Ejercicio 1: Pipeline de Limpieza y Reducción de Vocabulario

En este ejercicio aplicarás el flujo completo de limpieza y transformación de textos visto en esta unidad, sobre un subconjunto distinto de `data_corpus_inaugural`.

**Pasos sugeridos:**

1. Selecciona los **últimos 10 discursos** del corpus: `corpus_subset <- tail(data_corpus_inaugural, 10)`.
2. Tokeniza el corpus y calcula `ntoken()` y `ntype()` sobre la versión **sin procesar**.
3. Aplica, en orden: minúsculas → eliminación de puntuación y símbolos → eliminación de stopwords → stemming.
4. Vuelve a calcular `ntoken()` y `ntype()` sobre la versión procesada. ¿En qué porcentaje se redujo el vocabulario (*types*)?
5. Usa `kwic()` para buscar una palabra de tu elección (ej. `"freedom"`, `"war"`, `"economy"`) en la versión sin procesar y en la procesada. ¿Cambia el resultado o su interpretabilidad?
6. **Discusión**: ¿en qué casos esta reducción de vocabulario podría hacer perder información relevante para un análisis posterior con un modelo LLM (ver Unidad 6)?

> 💡 Tip: reutiliza el código de las secciones anteriores como plantilla. El objetivo es practicar el pipeline completo `corpus → tokens → limpieza` y reflexionar sobre sus efectos.
