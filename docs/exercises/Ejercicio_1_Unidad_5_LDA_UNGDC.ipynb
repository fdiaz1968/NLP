## Ejercicio 1: Topic Modeling con LDA sobre el UN General Debate Corpus

En la **Unidad 5a** ajustaste modelos **LDA** sobre el corpus de críticas de cine de Pang & Lee, probando distintos valores de $k$ y evaluando su calidad con *perplexity* y el método del codo.

En este ejercicio aplicarás el mismo flujo de trabajo a un corpus muy distinto: el **UN General Debate Corpus (UNGDC)**, más de 10.000 discursos de 202 países ante la Asamblea General de la ONU (1946–2023), disponible en Kaggle bajo licencia **CC0 (dominio público)** — el mismo corpus que usaste en el Ejercicio 1 de la Unidad 3.

🔗 [UN General Debate Corpus 1946-2023 — Kaggle](https://www.kaggle.com/datasets/namigabbasov/united-nations-general-debate-corpus-1946-2023)

### Pasos sugeridos

1. **Cargar el corpus.** Si ya construiste el código de carga en el Ejercicio 1 de la Unidad 3, reutilízalo (recuerda: la columna identificadora única es `doc_id`, no `country`). Trabaja con un subconjunto manejable (por ejemplo, un rango de años) para mantener los tiempos de ejecución razonables.
2. **Preprocesar con quanteda:** tokeniza, elimina *stopwords*, puntuación y números, construye la DFM y aplica `dfm_trim()` para controlar la sparsity — los discursos diplomáticos usan mucho vocabulario formulario (`"united"`, `"nations"`, `"assembly"`, `"secretary-general"`...) que conviene revisar y posiblemente agregar a la lista de *stopwords*.
3. **Ajustar modelos LDA** (`topicmodels::LDA`) con al menos **3 valores de $k$** (por ejemplo, 10, 15 y 20).
4. **Elegir un $k$**, justificando la decisión con *perplexity* y/o el método del codo, igual que en 5a.
5. **Inspeccionar y etiquetar manualmente** los tópicos del modelo elegido — ¿qué temas de política internacional emergen? (comercio, seguridad, cambio climático, derechos humanos, desarrollo...)
6. **Exportar** las palabras principales de cada tópico a un CSV con columnas `topic`, `term`, `beta`, `k` — el mismo formato que `lda_topterms_k*.csv` de la Unidad 5a. Lo necesitarás para el **Ejercicio 2**.

> 💡 **Tip:** a diferencia de las críticas de cine (cortas, coloquiales), los discursos de la ONU son largos y muy formales. Si los tópicos resultantes parecen dominados por palabras genéricas del género diplomático en vez de por temas específicos, es una señal de que necesitas ajustar tu lista de *stopwords* o tu `dfm_trim()`.
