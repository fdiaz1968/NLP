### Cómo obtener el corpus

Se obtiene directamente vía NLTK, igual que hiciste con `movie_reviews` en la Unidad 5d:

```python
import nltk
nltk.download("reuters")

from nltk.corpus import reuters

documentos = [reuters.raw(fileid) for fileid in reuters.fileids()]
print(f"Número de documentos: {len(documentos)}")
print(f"Número de categorías: {len(reuters.categories())}")
```
