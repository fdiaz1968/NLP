### Sobre el dataset

El dataset tiene ~11.900 tweets (9.938 entrenamiento / 2.486 validación), cada uno etiquetado como:

- `0` = **Bearish** (pesimista / expectativa de baja)
- `1` = **Bullish** (optimista / expectativa de alza)
- `2` = **Neutral**

### ⚠️ El problema del mapeo de etiquetas

Los modelos de la Unidad 6 predicen `positive` / `negative` / `neutral` — un juicio de **tono**, no de **dirección de mercado**. Este dataset, en cambio, usa `Bullish` / `Bearish` / `Neutral` — un juicio sobre hacia dónde se espera que se mueva el precio.

**Estos dos conceptos no son lo mismo.** Una noticia puede tener un tono claramente negativo ("la empresa despide al 20% de su plantilla") y aun así ser interpretada por el mercado como alcista (si los inversionistas leen el recorte de costos como señal de disciplina financiera). Antes de mapear `positive → Bullish` y `negative → Bearish`, tienes que decidir explícitamente si asumir esa equivalencia — y justificarlo, o proponer una alternativa.
