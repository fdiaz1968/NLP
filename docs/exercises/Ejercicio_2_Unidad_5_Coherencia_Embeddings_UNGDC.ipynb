## Ejercicio 2: Evaluando la Coherencia de Tópicos del UN General Debate Corpus con Embeddings

En la **Unidad 5c** mediste la coherencia semántica de los tópicos de LDA usando *embeddings* y similitud coseno, y en la **Unidad 5d** comparaste la distribución completa de coherencias entre distintos enfoques con un *boxplot* y una batería de tests estadísticos formales (Shapiro-Wilk, Levene, Welch, Mann-Whitney, mediana de Mood).

En este ejercicio aplicarás ese mismo método de evaluación al **UN General Debate Corpus**, usando los modelos LDA que ajustaste en el **Ejercicio 1**.

### Pasos sugeridos

1. **Cargar** el o los CSV que exportaste en el Ejercicio 1 (columnas `topic`, `term`, `beta`, `k`).
2. **Reutiliza** (o reimplementa) la función `coherencia_por_embeddings()` de la Unidad 5c: para cada tópico, toma sus 15 palabras más representativas, conviértelas en *embeddings* con `all-mpnet-base-v2`, y promedia la similitud coseno entre todos los pares de palabras.
3. **Calcula la coherencia** de cada tópico, para cada uno de los valores de $k$ que probaste en el Ejercicio 1.
4. **Visualiza** los resultados con un *boxplot* (estilo Unidad 5c/5d): cajas de un color, puntos individuales con *jitter*, y algún indicador de la tendencia entre los promedios de cada grupo.
5. **Aplica tests formales** para comparar las distribuciones entre los distintos valores de $k$. A diferencia de la Unidad 5d (que comparaba solo **dos** grupos con Welch/Mann-Whitney), aquí tienes **tres o más grupos** — necesitas un test que compare más de dos grupos a la vez, no una serie de comparaciones de a pares. Investiga qué test paramétrico y qué alternativa no paramétrica corresponden a esa situación.
6. **Responde:** ¿qué valor de $k$ recomendarías para modelar tópicos en discursos de la ONU? ¿El patrón de coherencia por $k$ se parece al que viste con las críticas de cine, o es distinto? ¿Por qué podría ser distinto, dado el tipo de texto (formal, largo, con vocabulario diplomático repetitivo)?

> 💡 **Tip:** el test que usaste en 5d para *dos* grupos no se generaliza directamente a *tres o más* — usarlo por pares (k=10 vs. 15, 10 vs. 20, 15 vs. 20) infla la probabilidad de error tipo I. Busca el test diseñado específicamente para comparar tres o más grupos de una vez.
